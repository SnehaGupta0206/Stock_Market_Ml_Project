### Data Cleaning And Preprocessing

```python
# Import the Libraries
import numpy as np
import pandas as pd
```

```python
# Load NIFTY50
nifty50 = pd.read_excel("../data/raw/Features NIFTY 50_ALL1.xlsx")
nifty50["Index"] = "NIFTY50"

# Load BANKNIFTY
banknifty = pd.read_excel("../data/raw/Features NIFTY BANK_ALL.xlsx")
banknifty["Index"] = "BANKNIFTY"

# Load MIDCAP
midcap = pd.read_excel("../data/raw/Features NIFTY MIDCAP_ALL.xlsx")
midcap["Index"] = "MIDCAP"

# Store datasets
datasets = {
    "NIFTY50": nifty50,
    "BANKNIFTY": banknifty,
    "MIDCAP": midcap
}
```

```python
# Check missing values and duplicate rows in each dataset
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("Missing values:\n", df.isnull().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
# One column has an extra space in its name: "Open _(t-1)" -> "Open_(t-1)"
for name, df in datasets.items():
    df.columns = df.columns.str.strip().str.replace(" _(t-1)", "_(t-1)")
    datasets[name] = df
print(nifty50.columns.tolist())
```

```python
# BANKNIFTY & MIDCAP store Date as text like 09-Nov-15 -> convert to datetime and sort
for name, df in datasets.items():
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")   # warning message is normal here
    df = df.sort_values("Date").reset_index(drop=True)
    datasets[name] = df
    print(name, "|", df["Date"].dtype, "|", df["Date"].min().date(), "->", df["Date"].max().date())
```

```python
# Basic rule: a day's HIGH can never be smaller than its LOW
for name, df in datasets.items():
    print(name, "-> rows where High < Low:", (df["High_(t-1)"] < df["Low_(t-1)"]).sum(), "of", len(df))

# BANKNIFTY breaks this rule in almost EVERY row -> file is wrongly made at source -> remove it
del datasets["BANKNIFTY"]
print("We continue with:", list(datasets.keys()))
```

```python
# Empty cells exist only in the first rows: indicators need a few days before they can exist
# (SMA_6 needs 5 past days, CMF needs ~19). We drop those first rows instead of filling fake values.
for name, df in list(datasets.items()):
    before = len(df)
    df = df.dropna().reset_index(drop=True)
    datasets[name] = df
    print(name, "| dropped", before - len(df), "warm-up rows | now", len(df), "rows | missing left:", df.isnull().sum().sum())
```

```python
# Save cleaned data (same style as class: index=False)
nifty50 = datasets["NIFTY50"]
midcap  = datasets["MIDCAP"]

nifty50.to_csv("../data/processed/nifty50_clean.csv", index=False)
midcap.to_csv("../data/processed/midcap_clean.csv", index=False)

# one combined file with both indices - needed for comparison charts in EDA
combined = pd.concat([nifty50, midcap], ignore_index=True)
combined.to_csv("../data/processed/all_indices_clean.csv", index=False)

print("Saved 3 clean files. Combined shape:", combined.shape)
```
